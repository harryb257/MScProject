```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from pathlib import Path

from sklearn.metrics import roc_curve, roc_auc_score
```

```python
merged_results_folder = Path(\
    '/Users/harry/Documents/Data Science MSc/PROJECT/MScProject/ESM2_pipeline_8M_conf_mat/HB/output_weighted/merged_results')

roc_csv_file = merged_results_folder / 'test_roc.csv'

output_dir = merged_results_folder / 'roc_auc_plots'
output_dir.mkdir(exist_ok=True)

test_roc_data = pd.read_csv(roc_csv_file)

pathogens = test_roc_data['pathogen'].unique()
modes = test_roc_data['mode'].unique()
size = test_roc_data['model size'].unique()

model_size_order = ['8M', '35M', '150M', '650M']

def plot_roc(dataset, pathogen, mode): 
    """
    Function to generate roc curve with auc score for a given pathogen and mode
    """

    subset =  dataset[(dataset['pathogen'] == pathogen) & (dataset['mode'] == mode)]

    fig, ax = plt.subplots(figsize=(6, 6))

    for model_size in model_size_order:
        if model_size not in subset['model size'].unique():
            continue

        model_subset = subset[subset['model size'] == model_size]

        y_true = model_subset['label']
        y_score = model_subset['probs']

        fpr, tpr, thresholds = roc_curve(y_true, y_score)
        roc_auc = roc_auc_score(y_true, y_score)

        ax.plot(fpr, tpr, label=f'{model_size} (AUC = {roc_auc:.3f})')

    ax.plot([0, 1], [0, 1], linestyle='--', color='grey')
    ax.set_xlabel('False Positive Rate')
    ax.set_ylabel('True Positive Rate')
    ax.set_title(f'ROC Curves \n {pathogen} — {mode}')
    ax.legend(title='Model Size')
    ax.grid(alpha=0.3)
    
    fig.tight_layout()

    return fig

for pathogen in pathogens:
    for mode in modes:
        fig = plot_roc(test_roc_data, pathogen, mode)

        output = Path(output_dir) / f'{pathogen}_{mode}.png'

        plt.savefig(
            output,
            dpi=300,
            bbox_inches='tight'
        )

        plt.close(fig)

        print('Saved plot:', pathogen, mode)
```
